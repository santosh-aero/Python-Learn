# Objects, Methods, and Attributes
__Everything__ in Python is an object!
***

<img src="./img/object.png" alt="Drawing" style="width: 100px;"/>

Objects are the fundamental building blocks of Python programs. 

* __Attributes:__ Values associated with a particular object
* __Method:__ Functions defined within a class that operate on objects

```python
# Accessing attributes:
object_name.attribute

# Invoking a method:
object_name.method()
```

***
## Example 1: Methods and Attributes of a complex number

```python
y = 2 + 3j
```

```python
type(y)
```

```python
# Accessing an attribute
y.real
```

```python
y.imag
```

```python
# Invoking a method
y.conjugate()
```

***
## Example 2: Methods and Attributes of a NumPy Array

```python
import numpy as np
x = np.array([[1,2,3],[4,5,6]])
x
```

```python
type(x)
```

```python
# Accessing an attribute
x.shape
```

Alternatively, we can use the `shape()` function in NumPy. You will encounter both of these approaches!

```python
# Using the np.shape() function
np.shape(x)
```

And we can also invoke methods on numpy arrays:

```python
# Invoking a method
x.sum()
```
