## The `help()` Function and `?`

The `help()` function is a built-in Python function that provides interactive help and documentation.

```python
help(abs)
```

Similar information can be obtained with the shorthand `?`:

```python
abs?
```

```python
abs(-4)
```

```python
import math
help(math)
```

***
## Shift + Tab
When you write code and place the cursor within a function or method call, pressing [Shift] and [Tab] together will display a pop-up window with the docstring and parameter details for that function.

```python
# Note: Shift + Tab will not work if running directly on Coursera platform
math.cos()
```

***
## Tab Completion
After typing the name of an object following the `.`, character hit the [Tab] key. Typing [Tab] brings up a list of available options.

```python
import numpy as np
x = np.array([[1,2,3],[4,5,6]])
```
