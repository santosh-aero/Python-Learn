## Creating Arrays

```python
import numpy as np
```

#### The **`numpy.array()`** function
You can create an array from a Python list or tuple by passing it as an argument to this function:

```python
my_list = [1,2,3,4]
my_array = np.array(my_list)
print(my_array)
```

```python
type(my_array)
```

```python
my_2d_array = np.array([[1,2,3],[4,5,6]])
print(my_2d_array)
```

#### The **`numpy.arange()`** function
Creates an array with regularly spaced values. You can specify the start, stop, and step values as arguments.

```python
arr = np.arange(1,10)
print(arr)
```

```python
arr_odd = np.arange(1,10,2)
print(arr_odd)
```

#### The **`numpy.linspace()`** function
Creates an array of evenly spaced values within a specified range. It takes 3 arguments: start, stop (ending value), and num (specifies number of evenly spaced values you want in the array).

```python
vector = np.linspace(0,1,5)
print(vector)
```

#### The **`np.ones()`** and **`np.zeros`** functions
You can create arrays filled with zeros or ones of a specified shape.

```python
zeros = np.zeros(5)
print(zeros)
```

```python
ones = np.ones((2,3))
print(ones)
```

## Efficient Operations

### Element-wise arithmetic operations
NumPy by default will use `+,-,*,/` to operate on arrays __*item-by-item*__:

```python
x = np.array([[1,2],[3,4]])
print(x)
```

```python
y = np.array([[2,4],[3,5]])
print(y)
```

```python
x+y
```

```python
x*y
```

### Element-wise functions
NumPy provides various mathematical functions that can be applied element-wise to arrays, including `np.sqrt()`, `np.exp()`, `np.sin()`, `np.cos()`, and more.

```python
z = np.array([1, 4, 9])
print(z)
```

```python
np.sqrt(z)
```

```python
np.exp(z)
```

### Why not use `math`?
