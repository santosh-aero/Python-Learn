You can cast or convert a variable from one data type to another using built-in functions specific to each data type. Some common examples (though there are many others) include:

* `str()`
* `int()`
* `float()`
* `bool()`
* `list()`

```python
x = 20
type(x)
```

```python
float(x)
```

```python
type(float(x))
```

But be mindful that the data type makes sense...

```python
y = int("Hello!")
```

```python
y = int('3')
y
```

## Example: User Input

<img src="./img/user.png" alt="Drawing" style="width: 150px;"/>

You can obtain input from the user using the built-in `input()` function. The `input()` function prompts the user to enter some text, and it returns the entered text as a string.

```python
import math
Radius = input("Enter the circle's radius:")
```

```python
Area = math.pi*Radius**2
```

Make sure to first covert to a numeric data type! (Remember: `input()` returns a string)

```python
Radius = float(input("Enter the circle's radius:"))
```

```python
Area = math.pi*Radius**2
print(Area)
```
