## Numeric Types

#### Integers (`int`)
Integers represent whole numbers without any decimal point, such as 1, -2, 0, 3845, etc.

```python
x = 42
type(x)
```

#### Floating Point Numbers (`float`)
Floating point numbers represent decimal numbers, such as 3.14, -2.5, etc. They are used to store numbers with fractional parts.

```python
pi = 3.14
type(pi)
```

```python
3.8e5
```

```python
y = 200.
type(y)
```

#### Complex Numbers (`complex`)
Complex data types represent complex numbers. A complex number is a number that comprises a real part and an imaginary part. It is expressed in the form a + bj, where a is the real part and b is the imaginary part.

```python
z = 3 + 4j
type(z)
```

***
## Boolean Type 
#### Boolean (`bool`)
Booleans represent truth values, either `True` or `False`. They are used to perform logical operations and control flow in programs.

```python
answer = True
type(answer)
```

```python
10 == 20
```

***
## None Type 
#### None (`NoneType`)
`None` is a special value that represents the absence of a value or a null value.

```python
my_var = None
type(my_var)
```

***
## Sequence Types

#### Strings (`str`)
Strings represent sequences of characters enclosed in single quotes (' ') or double quotes (" "). They are used to store textual data.

```python
greeting = "Hello"
type(greeting)
```

We can extract parts of the string with __indexing__:

<img src="./img/index.png" alt="Drawing" style="width: 400px;"/>

```python
greeting[0]
```

```python
greeting[4]
```

```python
greeting[-1]
```

```python
greeting[2:4]
```

#### Lists (`list`)
Lists are ordered collections of items enclosed in square brackets [ ]. They can contain elements of different data types and can be modified (__mutable__). For example, `[8, 3.14, 'green', True]`.

```python
my_list = [8,3.14,'green',True]
type(my_list)
```

Lists can also be indexed:

```python
my_list[2]
```

Because lists are __mutable__, they can be changed!

```python
my_list[3] = False
my_list
```

```python
my_list.append([1,2,3])
my_list
```

#### Tuples (`tuple`)
Tuples are similar to lists but are enclosed in parentheses ( ) instead of square brackets. Tuples are __immutable__, meaning their elements cannot be changed once assigned.

```python
my_tuple = (8,3.14,'green',True)
type(my_tuple)
```

```python
my_tuple[0]
```

```python
my_tuple[3] = False
```

#### Range (`range`)
Ranges represent an immutable sequence of numbers and is commonly used for iterating over a sequence of values or generating a series of numbers.

```python
my_range = range(10)
type(my_range)
```

```python
my_range
```

```python
for i in range(10):
    print(i)
```

***
## Mapping type

#### Dictionaries `dict`
Dictionaries are used to store key-value pairs enclosed in curly braces { }. Each element in a dictionary consists of a key and its associated value. For example:

```python
my_dictionary = {'name':'Jenny','age':22}
type(my_dictionary)
```

```python
my_dictionary['name']
```
