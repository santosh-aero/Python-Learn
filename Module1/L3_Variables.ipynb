# Variables and Assignments
***

## What Are Variables?

A __variable__ is a symbolic name that is a reference or pointer to a value or object.

```python
x = 42
```

<img src="./img/variable.png" alt="Drawing" style="width: 300px;"/>

Once it's been created, we can manipulate it and use it in mathematical expressions or other variable assignments:

```python
x + 2
```

```python
x/2
```

Variables aren't restricted to numeric values. For example, they can hold lists, strings, and more. Here is an example of a string:

```python
greeting = "Hello!"
print(greeting)
```

Python is a **dynamically typed** language, which means we don't need to declare variables in advance. The `type()` function returns the type of the specified object. For example:

```python
type(x)
```

```python
type(greeting)
```

We can also override the variable by redefining it:

```python
x = 5
print(x)
```

***
## Naming Rules and Conventions

Names must start with a letter or underscore (not a number or other symbol):

```python
my_var = 8  # Is ok
```

```python
1my_var = 8   # Won't work!
```

Most special characters won't work:

```python
my_$var = 8
```

Avoid reserved words (e.g. `for`, `if`, `return`):

```python
for = 10
```

Python is *case sensitive*:

```python
x = 10
```

```python
X = 20
```

```python
x
```

```python
x == X
```

Best to use short/meaningful names (often lowercase, multiple words connected by underscore):

```python
my_var = 9
```

```python
MyVaRiAbLe = 9   # Not recommended
```

```python
My_variable_that_describes_the_radius_of_the_circle_I_need = 10   # Not recommended!
```

Check out the __PEP8__ style guide document for much more details! (But this is a great start...)  
https://peps.python.org/pep-0008/

## Additional Considerations

Variables act as references to the memory locations where the actual data is stored. Let's talk about what happens when a variable is assigned to a value (object):

```python
height = 10
width = 5
```

Value is stored in computer memory:

<img src="./img/capture.png" alt="Drawing" style="width: 300px;"/>

The assignment "binds the name to a value". 

```python
height*width
```

We can override the variables by redefining them (Python silently handles the cleanup through "garbage collection"):  
<img src="./img/override.png" alt="Drawing" style="width: 300px;"/>

```python
height = 7
```

```python
height*width
```

We can also define two variables which point to the same object in memory:  
<img src="./img/copy.png" alt="Drawing" style="width: 400px;"/>

```python
x = [1,2,3]
y = x
```

```python
y
```

```python
id(x)
```

```python
id(y)
```
